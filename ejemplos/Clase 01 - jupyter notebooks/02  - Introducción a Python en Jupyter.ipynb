### Diccionarios

Los diccionarios permiten manipular un par clave:valor (key:value) de forma muy flexible, siendo parte de la base de muchas implementaciones internas de Python. Se crean usando { } (llaves), donde cada elemento (clave:valor) se separa con una , (coma):

````python
numeros = {'tres':4,'uno':1, 'dos':2 }
print(numeros)
````

Los elementos son accedidos y modificados a través su índice, tal como en las listas, exceptuando que no hay un 
orden que comience desde el cero, sino que desde una clave presente en el diccionario.

````python
# Acceder a un valor a través de su clave
numeros['tres']
````

Nuevos elementos pueden ser agregados al diccionario usando la misma sintaxis del índice:

````python
# Agregar un nuevo par clave:valor al diccionario 'numeros'
numeros['perro'] = 12 
print(numeros)
````

**Nota**: Ten en cuenta que el orden en un diccionario no es relevante, fue diseñado pensando así. Esto permite que el acceso a sus elementos sea muy eficiente, independiente de su tamaño. Puedes ver información sobre los métodos aplicables a diccionarios en la [documentación oficial de python](https://docs.python.org/3/library/collections.html).

### Conjuntos
La última colección básica de elementos es el conjunto o `set`, siendo una colección no ordenada de elementos únicos. Es mutable: una vez que se ha creado un conjunto, puede ser modificado. Se definen como las listas y tuplas, pero usan llaves como los diccionarios:

````python
primos = {2, 3, 5, 7}
impares = {1, 3, 5, 7, 9}
````

Si recordamos la notación y propiedades de los conjuntos en la mátemática, existen operaciones como la unión, intersección, diferencia, diferencia simétrica, entre otras. Python nos brinda la posibilidad de utilizar métodos u operadores para realizarlas:

````python
# union: items appearing in either
# primos | impares      # # con un operador
primos.union(impares) # equivalente con un método
````

````python
# intersección: elementos que están en ambos grupos
# primos & impares             # con un operador
primos.intersection(impares) # equivalente con método
````

````python
# diferencia: elementos en primos pero no en impares
primos - impares           # con un operador
primos.difference(impares) # equivalente con método
````

````python
# diferencia simétrica: elementos que aparecen en un solo conjunto
primos ^ impares                     # con un operador
primos.symmetric_difference(impares) # equivalente con método
````

## Estructuras de control
En Python no hay llaves ({}), ni `begin`/`end` para marcar el comienzo y fin de un bloque, sino que eso se logra con la **indentación**. La indentación por defecto son 4 espacios en blanco, que serán necesarios para utilizar sentencias de control como `if`, `for`, `while` o definir funciones.

### If
El `if` es como el de otros lenguajes, pero no pide paréntesis y termina con : . Su sintaxis es:

```
if condicion :
    cuerpo del if (indentado con 4 espacios)
else:
    cuerpo del else (indentado con 4 espacios)
```


````python
# Chequear signo de número

x = 1
#if x > 0:
#    print("El número ingresado ("+str(x)+") es positivo.")
#    print("Esto es parte del if")
#else:
#    print("El número ingresado ("+str(x)+") es negativo.")
#    print("Esto es parte del else")

if x == 0:
    print("El número ingresado ("+str(x)+") es cero.")
elif x > 0:
    print("El número ingresado ",x," es positivo.")
else:
    print("El número ingresado ("+str(x)+") es negativo.")
````

### For
El `for` es una sentencia de control que repite la condición establecida en un comienzo. A aquella sentencia que repita condiciones se le llama *bucle* o *ciclo*. Tiene la siguiente sintaxis:
```
for variable in lista:
    cuerpo del for
```

En este caso, `variable` es la variable que va a ir cambiando, y `lista` es una lista de python (o un `iterable` que es parecido).

````python
print("- Impresión de los elementos de la lista:")

# Imprimir los strings de mi_lista por separado

mi_lista=["img","python","numpy", "pandas"]

for s in mi_lista:
        print(s, end=' ') # imprimir todo en la misma línea
````

````python
#Calcular la suma de los números e imprimirla

suma=0
pixel=[0.6,0.3,0.4]

for numero in pixel:
    suma+=numero # recordar que es lo mismo que suma = suma + numero
    print(suma)

print("- La suma de los números es:")
print(round(suma,3))

print("- El promedio de ellos es:")
print(round(suma/len(pixel),3))
````

Cuando no tenemos una lista y queremos hacer un `for` "común" y que la variable que cambia sea un número que va incrementándose, podemos utilizar la función `range`.

````python
# Realizar un for de 0 a 3, que imprima esos valores
print("- Un for de 0 a 3")
for i in range(4):
    print(i, end=' ')
````

En Python los índices comienzan en 0, y por eso los rangos también. Aún así, se puede comenzar el rango en otro valor en lugar de 0

````python
print("- Un for de 2 a 5:")
for j in range(2,6):
    print(j, end=' ')
````

### While
La sentencia `while` permite repetir la ejecución de un grupo de instrucciones mientras se cumpla una condición (es decir, mientras la condición tenga el valor `True`). La sintaxis del ciclo while es la siguiente:

```
while condicion:
    cuerpo del ciclo
```

````python
# Ejemplo 1 - while
i = 1
while i <= 50:
    print('- Valor de i antes de cambiar:', i)
    i = 3*i + 1
    print('- Valor nuevo de i:', i)
print("Programa terminado")
````

````python
# Ejemplo 2 - while
numero = int(input("Escriba un número positivo: "))
while numero < 0:
    print("¡Ha escrito un número negativo! Inténtelo de nuevo") # se muestra el error
    numero = int(input("Escriba un número positivo: ")) # se vuelve a preguntar la condición inicial al final
print("¡Este sí es un número positivo!")
````

**Nota**: Si la condición de un ciclo se cumple siempre, no terminará nunca de ejecutarse y tendremos lo que se denomina un *ciclo infinito*. Los ciclos infinitos no intencionados deben evitarse, ya que implica perder el control del programa. Para interrumpirlo, hay que pulsar la combinación de teclas `Ctrl+C`.

### Funciones
Las funciones se definen con la palabra clave `def` y tienen la sintaxis `def nombre_funcion(parametros):`. Para devolver un valor utilizamos la palabra clave `return`.

````python
# Esta funcion recibe dos números y devuelve su suma

def sumar(a,b):
    return a+b

c=sumar(2,5)
print("2 + 5 =", c)
print(sumar(8,9))
````

````python
# Esta funcion recibe una lista y devuelve la suma de sus elementos
def sumar_todos(lista):
    suma=0
    for v in lista:
        suma+=v
    return suma

mi_lista=[10,20,30,50]

# Se pueden `llamar` funciones en los argumentos de otra, como dentro de 'print', con el fin de
# no tener que alojar el espacio en memoria si no es necesario conservarlo.
print("- Los elementos de la lista suman:", sumar_todos(mi_lista))
````

## Ejercicio 7 - Frecuencia/ocurrencia de un valor
Escribir una función que reciba una lista y un valor y devuelva la cantidad de veces que aparece ese valor en la lista



````python
l=[1,4,2,3,5,1,4,2,3,6,1,2,1,3,5,1,1,5,3,2]
v=2

####################
#IMPLEMENTAR

# para 'v=2' debiese entregar 4 ocurrencias
````

````python
def ocurrencias(lista,valor):
    # IMPLEMENTAR
    freq = 0
    for i in range(0,len(lista)):
        if (lista[i] == valor):
            freq+=1   
    return freq 

##########################################

l=[1,4,2,3,5,1,4,2,3,6,1,2,1,3,5,1,1,5,3,2]
v=2

# para 'v=2' debiese entregar 4 ocurrencias
print('- Cantidad de ocurrencias para v='+str(v)+':', ocurrencias(l,v))
````
