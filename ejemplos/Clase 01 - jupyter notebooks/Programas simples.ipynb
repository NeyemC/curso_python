# Números primos

##  Objetivo

Determinar si un número es primo o compuesto:
    Un número primo es aquel que solo es divisible por uno o por si mismo.
    Divisible: cuando el resto de división por un número dado es 0.
        Resto =  Dividendo  (divisor x cuociente) o
        Dividendo - Resto = divisor x cuociente
        “Lo que hay que quitar al dividendo para que el número sea divisible”        

**Algoritmo**: Lenguaje natural: (Criba de Eratóstenes)
    Un número 𝑛 es primo si no podemos encontrar un número entero entre 2 y 𝑛−1 que lo haga divisible por dicho 
    número.


```python
n = int(input('Ingrese n: '))
divisor = 2
es_primo = True
while (divisor != n):
    if (n % divisor == 0) :
        es_primo = False   
        divisor = n
    else:
        divisor = divisor + 1
if (es_primo == True) :
    print (n, 'es primo')
else:
    print (n, 'es compuesto')
```

# Años bisiestos: if else vs if elif ... else

**Algoritmo**: Lenguaje natural:
    Un año es bisiesto cuando es divisible por **4**, **con excepción** de que sea divisible por 100, **pero si** 
    cuando lo es por **400**

## Solución 1 con if else

```python
#Solución 1 con if else
agno = int(input('Ingrese agno: '))
if agno % 4 == 0 :
    if agno % 100 == 0 and agno % 400 != 0 :
        bisiesto = False
    else :
        bisiesto = True
else:    
      bisiesto = False
print (bisiesto)
```

## Solución 2 con if elif else

```python
#Solución 2 con if elif else
agno = int(input('Ingrese agno: '))
if agno % 400 == 0 :
    bisiesto = True
elif agno % 100 == 0 :
    bisiesto = False
elif agno % 4 == 0 :
    bisiesto = True
else:    
      bisiesto = False
print (bisiesto)
```

## Solución 3 con if else y operadores lógicos

```python
#Solución 3 con if else y operadores lógicos
agno = int(input('Ingrese agno: '))
if (agno % 4 == 0 and agno % 100 != 0) or agno % 400 == 0 :
   bisiesto = True
else :    
     bisiesto = False
print (bisiesto)
```
