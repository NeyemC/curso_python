### Acerca del datatype `object`

Pandas almacena, en caso de no ser definido explícitamente, los objetos de tipo `string` como `object`. Una forma de explicitar el datatype puede es a través de la API de numpy:

```python
import numpy as np
x = np.array(['Testing', 'a', 'string'], dtype='|S7') # usa un datatype explícito de numpy: string de largo 7
y = np.array(['Testing', 'a', 'string'], dtype=object) # usa un objeto genérico de python
print(x)
print(y)
```

El primer caso presenta ventajas en eficiencia, pero desventajas en versatilidad:

```python
x[1] = 'a really really really long' # almacenar un string muy largo en el segundo elemento
x
```

```python
y[1] = 'a really really really long' # el datatype object permite almacenar strings de largo arbitrario
y
```

¿Por qué es esto importante?

La eficiencia de pandas y numpy depende principalmente de dos cosas:
- Estructuras predefinidas en C como `int64`, orientadas a operaciones eficientes
- Pandas aloja estas estructuras en espacios contiguos de la memoria, permitiendo acceso rápido a ellas.

![numpy-object](figures/numpy-object.png)

- Un elemento tipo `object` en un dataframe contiene un *puntero* (una dirección en memoria) hacia un objeto en python, __el cual no necesariamente es contiguo__. 


- Pensar en una guía de teléfonos. Para un computador, un dataframe con objetos del tipo `int64` es como si un humano leyera los números de teléfono por ciudad, empezando por de menor a mayor.

        - 0123456, Pedro Pérez, Temuco
        - 0123457, Pedro Peter, Temuco

    Si estos se guardaran como `object`. un computador los verá como si una entrada de teléfonos fuera un índice:

        - Teléfono de Pedro Pérez, Temuco: Ver página 45
        - Teléfono de Pedro Peter, Temuco: Ver página 156

TL;DR

- Es importante evitar en lo posible la presencia de objetos de tipo `object`, pues reducen el potencial de Pandas.
- En datasets muy grandes esto se vuelve crítico

> Series and Index are equipped with a set of string processing methods that make it easy to operate on each element of the array. Perhaps most importantly, **these methods exclude missing/NA values automatically**. These are accessed via the str attribute and generally have names matching the equivalent (scalar) built-in string methods

Categorizar la columna `genre`, utilizando lo que se conoce como _one hot encoding_ (no confundir con _label encoding_)

Qué es _one hot encoding_?

### *Label encoding* y *One hot encoding*

 Una variable categórica se refiere a las categorías de un elemento, pudiendo ser

- Color de un auto: rojo, negro, blanco
- Estación del año: Verano, Invierno, Primavera, Otoño

Estas variables son fáciles de distinguir por humanos, pero computacionalmente difíciles y/o costosas para un computador. La idea es convertir estas variables en números

![label-encoding](figures/label-encoding.jpeg)

```python
movies["genres"].str.get_dummies(sep='|').head() # one hot encoding sobre columna "genres"
```

Notamos que existen 24 géneros de películas distintas. Finalmente, unimos ambos dataframes

https://pandas.pydata.org/pandas-docs/stable/merging.html

```python
one_hot_encoding = movies["genres"].str.get_dummies(sep='|') # guardamos el encoding en un nuevo dataframe
```

```python
movies = pd.concat([movies, one_hot_encoding], axis=1)
movies.head()
```

¿Qué películas son de acción 🔫 y romance 💕 que duren más de 2 horas?

```python
movies[(movies["Action"] == 1) & (movies["Romance"] == 1) & (movies["runtime"] > 120)].head()
```
