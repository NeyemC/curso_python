Numpy en si mismo es un tema muy grande, por lo que se intentará dar énfasis en los siguientes puntos:

__¿Por qué usar Numpy?__

- Permite operaciones vectorizadas sobre arreglos (arrays) para limpieza y manipulación de datos, filtrado, transformación, y otro tipo de operaciones
- Implementación nativa de algoritmos comunes como ordenamiento (sorting), búsqueda (search), y operaciones propias de conjuntos, tales como:
  - Extraer valores únicos
  - Categorizar valores no categóricos
- Implementa métodos para estadística descriptiva simple, así como herramientas para agrupar/sintetizar datos: promedios, sumas, cuentas, etc.
- Manipulación de datos para tareas de fusión (merging) y unión (join)
- Expresión de operaciones lógicas ("_si ocurre esto, entonces haz esto_") en términos de operaciones vectorizadas, en vez de operaciones iterativas como bucles `for` y `while`.
- Manipulación de datos agrupados

## Comentarios acerca de la escalabilidad (Big-O notation y por qué es importante Numpy)

Al hablar de la eficiencia de numpy es inevitable mencionar algunas nociones básicas sobre escalabilidad. En ciencias de la computación, se utiliza lo que se conoce como _Big-O Notation_ para describir la manera en que el número de operaciones requeridas por un algoritmo _escala_ (aumenta en tamaño) en la medida que aumenta la información con la que opera. 

Big-O notation, de modo simple y sin querer profundizar en contenidos de ciencias de computación, nos dice cuánto tiempo se demora un algoritmo en procesar un _dataset_, y nos da una noción de cómo este tiempo aumenta a medida que el _dataset_ aumenta en tamaño. 


Por ejemplo, si se tiene un algoritmo A, de orden $n$ o $O(n)$ , esto quiere decir que si este algoritmo se demora 1 segundo en procesar mil líneas de texto ($n=1.000$), entonces es esperable que este mismo algoritmo se demore 5 segundos en procesar $n=5.000$.

Por otro lado, si se tiene un algoritmo B, cuyo orden es de $O(n^2)$, el cual demora 1 segundo en procesar $n=1000$, entonces es esperable que este algoritmo se demore 25 segundos en procesar $n=5.000$ datos. 

En general en el procesamiento de datos se asocia el valor de $n$ al tamaño del dataset. Es de esta manera que si se considera el análisis de un dataset muy grande, digamos del orden de trillones de datos, entonces la diferencia entre $O(n)$ y $O(n^2)$ se vuelve de suma importancia.

Considerar además de la notación _Big-O_ no habla específicamente del tiempo _cronometrado_ que dura la ejecución de un algoritmo, sino más bien de cómo este algoritmo escala en la medida que crece $n$. Se considera entonces que un algoritmo $O(n)$ escala __mejor__ que uno $O(n^2)$, pero para dataset pequeños esta diferencia puede no ser muy notoria.

Es importante entender los órdenes de complejidad pues permite establecer comparaciones.

## Datatypes en `ndarrays`

Un *datatype* es un objeto especial que contiene información acerca de los datos que almacena un arreglo. Estos poseen características tales como

- Cantidad de bits usados para representar un número: esto influye en los números máximos posibles de representar usando un data type en particular, y se especifican generalmente con la cantidad como sufijo i.e. `int8` es de tamaño 8 bits, y `float128` es de tamaño 128 bits.

- Presencia de signo (_unsigned_ (`uint8`) versus _signed_ (`int8`))

- Si es decimal (float versus integer)

````python
# ejemplos de data types
import numpy as np
# enteros
int_types = ["int8", "uint8", "int16", "uint16"] #tipos de enteros en NumPy
for x in int_types:
    print(np.iinfo(x))
````

````python
import numpy
# ejemplos de tata types flotantes
float_types = ["float16", "float32", "float64"]
for x in float_types:
    print(np.finfo(x))
````

Notar el campo de `precision`. ¿Qué significa? ¿Cómo varían los campos `min` y `max`?

Esto permite a numpy realizar optimizaciones especiales según el dtype siendo usado. Estas optimizaciones pueden ser bien de uso de memoria o velocidad.

### Indexación y extracción de subconjuntos de arreglos

`ndarray` se comporta en gran manera como listas en python:

````python
arr = np.arange(10)
print(arr)
````

````python
print(arr[5])
print(arr[5:8])
````

````python
arr[5:8] = 12 # asignar 12 a los valores en el rango 5-8
print(arr)
````

Una diferencia importante, sin embargo, es que subconjuntos de un `ndarray` son realmente _vistas_ del ndarray original.  Esto significa que la extracción de subconjuntos no produce una copia de los datos, y que *cualquier modificación en el subconjunto se verá reflejado en el conjunto original y viceversa.

Dado que numpy está diseñado para trabajar con _grandes_ conjuntos de datos, es posible imaginar los problemas de rendimiento resultantes de estar copiando constantemente los datos.

````python
# extraer un subconjunto
arr_slice = arr[5:8]
print(arr_slice)
print(arr)
````

````python
# reemplazar un solo valor en el subconjunto
arr_slice[1] = 12345
print(arr) # ver cambio en arreglo original
````

````python
arr_slice[:] = 64 # reemplazar todos los valores en el subconjunto con el valor 64
print(arr) # ver cambios en arreglo original
````

Si se desea una copia explícita, se debe usar `copy()`:

````python
arr_slice_copy = arr_slice.copy() # hacer una copia del subconjunto anterior
print(arr_slice_copy)
arr_slice_copy[:] = 0 # reemplazar todos los valores por cero
print(arr_slice_copy) # 
print(arr) # cambios no son reflejados!
````

## Indexación lógica

La indexación lógica implica la utilización de máscaras o *masking*. Una máscara es un arreglo auxiliar que se utiliza para indexar otros arreglos.

Es posible extraer subconjuntos de un arreglo en base a condiciones lógicas. Una condición lógica puede ser, por ejemplo

- `age > 28` (Mayor a)
- `name == "Manuel"` (Igual)
- `name != "Manuel"` (No igual)

````python
names = np.array(['Bob', 'Joe', 'Will', 'Bob', 'Will', 'Joe', 'Joe']) # produce un arreglo de 1x7
data = np.random.randn(7, 4) # produce una matriz de 7x4 de números aleatorios
````

````python
print(names)
````

````python
print(data)
````

````python
names == 'Bob' # coindice en los índices 0 y 3
````

````python
data[names == 'Bob']
````

El arreglo resultante es aquel en los cuales `names == "Bob"` es `True`, es decir, las filas 0 y 3

````python
data[names == 'Bob', 2:]
````

````python
data[names == 'Bob', 3]
````

````python
names != 'Bob' # != significa "no es igual a"
data[~(names == 'Bob')] # o alternativamente, '~' es un operador NOT que niega (invierte) los resultados
````

````python
cond = names == 'Bob' # cond es una máscara
print(cond)
````

````python
data[~cond] # extrae los valores de data en los cuales se cumpla cond == False (notar el ~)
````

Es posible utilizar lógica binaria (o booleana) para construir máscaras complejas

````python
#['Bob' 'Joe' 'Will' 'Bob' 'Will' 'Joe' 'Joe']

mask_1 = (names == 'Bob') 
mask_2 = (names == 'Will')
print(mask_1,"\n")
print(mask_2)
````

````python
mask = mask_1 | mask_2 # '|' (pipe) es equivalente a un OR
print(mask)
````

Notar en en numpy los operadores lógicos `or` y `and` no funcionan, y deben usarse `|` y `&` en reemplazo.

````python
#[[ 0.75836315 -0.66052433  0.86258008 -0.0100319 ]
# [ 0.05000936  0.67021559  0.85296503 -0.95586885]
# [-0.02349332 -2.30423388 -0.65246884 -1.21830198]
# [-1.33260971  1.07462269  0.72364151  0.69000185]
# [ 1.00154344 -0.50308739 -0.62227423 -0.92116861]
# [-0.72621349  0.22289555  0.0513161  -1.15771947]
# [ 0.81670694  0.43360961  1.01073695  1.82487521]]


data[mask]
````
