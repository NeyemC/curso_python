# Objetos fundamentales en Pandas

De forma simple, los objetos elementales de pandas son tres: `Series`, `Dataframes` e `Index`. En estos objetos, las filas y columnas se identifican con _etiquetas_ (*labels*) en vez de sólo índices numéricos, como ocurre en Numpy.

¿Cómo accedemos a estas librerías?

````python
import numpy as np
import pandas as pd
````

Es convención importar estas librerías usando los alias `np` y `pd`.

## El objeto `pd.Series`

Una serie en pandas es un arreglo __unidimensional__ de datos indexados. Una forma de crearlos es:

````python
data = pd.Series([0.25, 0.5, 0.75, 1.0])
data # equivalente a print (data)
````

Es posible acceder a los valores directamente, los cuales son representados internamente como un arreglo de `numpy`:

````python
data.values
````

````python
type(data.values)
````

El objeto ``index`` es un arreglo del tipo ``pd.Index``

````python
data.index
````

Al igual que un arreglo de Numpy, es posible indexar una serie de Pandas

````python
data[3]
````

````python
print(data[0:2])
````

### ``Series`` como una forma generalizada de un arreglo NumPy

La diferencia principal entre estructuras de Pandas y NumPy es la presencia explícita del `index`.

Esto le entrega a `pd.Series` algunas características adicionales: Por ejemplo, los índices pueden ser no sólo enteros sino que también letras:

````python
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=['a', 'b', 'c', 'd'])
data
````

````python
data['c'] # es posible acceder a través del indice
````

Es posible además usar índices no contiguos o no secuenciales:

````python
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=[2, 5, 3, 7])
data
````

### `Series` como `dict`

Es posible ver un objeto tipo `Series` como una variante de un diccionario de python. La diferencia está en el código que está detrás: de la misma manera que numpy resulta más eficiente que utilizar una lista `list()` en python, un objeto del tipo `Series` es más eficiente que `dict()`

Esta analogía ``Serie`` vs ``dict()`` puede resultar más clara si se construye el primero a partir del segundo:


````python
dicc_poblacion = {'California': 38332521,
                   'Texas': 26448193,
                   'New York': 19651127,
                   'Florida': 19552860,
                   'Illinois': 12882135}
````

````python
poblacion = pd.Series(dicc_poblacion) # convertir a Series
poblacion
````

Por defecto el `index` de una serie se obtiene de las llaves (`dict.keys()`) del diccionario. Desde aquí es posible realizar operaciones como si fuera un objeto `pd.Series`:

````python
poblacion['California']
````

Sin embargo y a diferencia de un diccionario, `Series` introduce mejoras en la indexación:

````python
poblacion['California':'New York'] # indexa un rango
````

### Construyendo dataframes

Antes de comenzar, es importante mencionar que un dataframe no persiste operaciones implícitamente, es decir muchas operaciones **no modifican** el dataframe en memoria. Esto es igual en python en general. Consideremos el siguiente ejemplo:

```python
my_dummy_variable = 10
my_dummy_variable + 100
print(my_dummy_variable)
```

Qué valor tendrá `my_dummy_variable`?

````python
my_dummy_variable = 10
my_dummy_variable + 100
print(my_dummy_variable)
````

En el caso de dataframes:

````python
df = pd.DataFrame({'a':[1,2], 'b':[3, 4]})
df
````

````python
df['a'] * 10 # esta operación no modifica df
#print (df['a'] * 10)
df
````

¿Cómo persistimos los cambios?

````python
df['a'] = df['a'] * 10 # es necesario reemplazar un elemento con otr para persistir los datos
df
````

Algunos métodos de pandas poseen el argumento `inplace=False` por defecto. Cambiarlo a `True` en general se considera **mala práctica**.

## El objeto `pd.Index` de Pandas

Hemos visto que ``Series`` y ``DataFrames`` comparten un índice explícito que permite referenciar y modificar datos.

Como ejemplo, construyamos un índice usando una lista de enteros:

````python
ind = pd.Index([2, 3, 5, 7, 11])
ind
````

Objetos tipo ``Index`` tienen también atributos de numpy:

````python
# size   : Return the number of elements in the underlying data.
# shape  : Return a tuple of the shape of the underlying data.
# ndim   : Number of dimensions of the underlying data, by definition 1.
# dtype  : Return the dtype object of the underlying data.

print(ind.size, ind.shape, ind.ndim, ind.dtype)
````

Una diferencia importante es que los objetos tipo índice son **inmutables**: no pueden ser modificados:

````python
ind[1] = 0 # arroja un error
````

Esto ofrece una capa de seguridad para compartir índices entre distintos dataframes, y evitar cambios no advertidos

## Selección e indexado

Al igual que un diccionario, el objeto `Series` provee un mapeo _(key -> value)_

````python
# Ver como referencia: https://pandas.pydata.org/pandas-docs/stable/reference/series.html
import pandas as pd
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=['a', 'b', 'c', 'd'])
data
````

````python
# verificar que un índice existe en el objeto
'a' in data 
````

````python
# obtener las keys del objeto
# key: Alias for index.
data.keys() 
````

````python
# items (): Lazily iterate over (index, value) tuples
# utilizar el método items() para obtener tuplas (key, value)

for key, value in data.items():
    print(key, '->', value) 
````

````python
# o en una línea
list(data.items())
````

````python
data['e'] = 1.25 # asignar valores referenciando keys, se agrega un valor a un índice inexistente anteiormente
data
````

````python
# indexando a través de un índice explícito
data['a':'c'] # notar que el tercer índice sí aparece, a diferencia de los rangos en los arreglos.
````

````python
# indexando a través de las posiciones de index
data[0:2] # notar que el tercer índice *no* aparece, es consistente con el comportamiento de python y NumPy()
````

````python
# masking, como se vio en NumPy
# Despliga todos los valores de data que sean mayores 0.3 y menosres de 0.8
# (data > 0.3) & (data < 0.8) arroja una lista de valores booleanos, que es donde la condición es verdadera.

data[(data > 0.3) & (data < 0.8)]
````

````python
# fancy indexing

data[['a', 'e']]
````

### Métodos propios de indexación: loc, iloc, and ix
`Series.loc` 	Access a group of rows and columns by label(s) or a boolean array.

`Series.iloc` 	Purely integer-location based indexing for selection by position.

````python
data = pd.Series(['a', 'b', 'c'], index=[1, 3, 5])
data
````

Puede producirse una confusión al indexar directamente:

````python
# Explicito, índice hace mención al valor del índex=2
data[5]
````

````python
# Implicito, índice hace mención al valor como si se tratara de una lista en python
data[1:3] # Nótese que indiza usando < en el rango derecho
````

A continuación vemos que el primer ejemplo se refiere al índice explícito, mientras que el segundo se refiere al mismo de manera implícita, como si se tratara de una lista de python.

Debido a estas confusiones __Pandas__ provee métodos especiales de indexación: primero `loc` permite referenciar siempre al índice de manera __explícita__. Esto es, referenciando a los valores del índice de Pandas.

````python
data.loc[1]
````

````python
# Nótese que usando data[1:3] dió un resultado diferente. Aquí usa el valor del índice y no la posición
data.loc[1:3] 
````

Existe además ``iloc``, el cual permite referenciar el índice siempre de manera **implícita**. Esto es tratando las filas como si se tratara de una lista de python.

````python
data.iloc[1]
````

````python
data.iloc[1:3] # se pregunta por los rangos de posición dentro de la Serie
````

Un tercer atributo, ``ix``, resulta como un híbrido de los dos anteriores, y para el caso de `Series` es equivalente a acceder a los datos a través de `[]`. Su propósito tiene más sentido en el contexto del trabajo con Dataframes, aunque su uso se desaconseja, teniendo las últimas versiones de python una [advertencia al respecto](http://pandas.pydata.org/pandas-docs/stable/indexing.html#ix-indexer-is-deprecated):

````python
data.ix[3] 
````

### Indexación en dataframes

````python
area = pd.Series({'California': 423967, 'Texas': 695662,
                  'New York': 141297, 'Florida': 170312,
                  'Illinois': 149995})

pop = pd.Series({'California': 38332521, 'Texas': 26448193,
                 'New York': 19651127, 'Florida': 19552860,
                 'Illinois': 12882135})

# Creamos un dataframe compuesto de dos series que comparten el mismo índice, cuyas columnas se llaman 'area' y 'population'
data = pd.DataFrame({'area':area, 'population':pop})
data
````

Es posible acceder a las columnas de dos formas:

````python
# Accede a la columna área
data['area'] 
````

````python
data.area # alternativamente
````

````python
data.area is data['area']
````

Siempre es mejor utilizar el primer método, debido a que el segundo puede a veces conflictuar con métodos propios de un dataframe. (Por ejemplo si un dataframe tiene una columna llamada `shape`). 

En lo posible evitar hacer asignaciones del tipo `data.pop = z`, y preferir siempre la alternativa `data['population'] = z`

````python
data.pop is data['population']
````

Es posible asignar nuevas columnas en función de las existentes:

````python
# Crea una nueva serie llamada 'density' dentro del DataFrame con un cálculo de las columnas 'population' y 'area'
data['density'] = data['population'] / data['area']
data
````

También es posible transponer la tabla:

````python
data.T # Muestra la data traspuesta, pero no la modifica, para hacerlo hay que asignar data = data.T
````

### ``loc``, ``iloc`` en Dataframes
Recordar: 
`loc` -> explícito;
`iloc` -> implícito

````python
data.iloc[:3, :2] # hasta índice <3 y columna <2
````

````python
data.loc[:'Illinois', :'population'] # indíce hasta illinois, columnas hasta pop (inclusive)
````

````python
# Si hacemos esto explota porque el DataFrame no tiene valores de índices enteros, son string
data.loc[:3, :2] # hasta índice <3 y columna <2
````

````python
data.loc[data["density"] > 100, ['population', 'density']]
````

````python
data.iloc[0, 2] = 0 # reemplazar valores en la primera fila, tercera columna (0,2)
data
````

## Aplicación de funciones con `apply` y `ufunc`

Conocidas como `ufuncs` in Numpy, también son aplicables a dataframes

````python
# numpy.random.randn(d0, d1, ..., dn)¶
# Return a sample (or samples) from the “standard normal” distribution.

frame = pd.DataFrame(np.random.randn(4, 3), columns=list('bde'),
                     index=['Utah', 'Ohio', 'Texas', 'Oregon'])
frame
````

````python
np.abs(frame) # valor absoluto
````

````python
# soporte de funciones anónimas
# https://thepythonguru.com/python-lambda-function/

f = lambda x: x.max() - x.min()  # resta el maximo del minimo de cada arreglo
frame.apply(f) # Por defecto toma el eje de las filas

# recordar que estamos tomando f como el DataFrame que tiene valores negativos 2 celdas arriba.
````

````python
frame.apply(f, axis='columns') # aplicado sobre el eje de las columnas
````

Recordar 💡

- `axis=0` -> filas: Método colapsa *filas* -> Quedan sólo columnas
- `axis=1` -> columnas: Método colapsa *columnas* -> Quedan sólo filas 
    

````python
# también es posible utilizar funciones definidas
# Esto toma un dataframe x
# calcula [x.min(), x.max()]
# regresa un objeto pd.Series con nuevos índices

def defined_function(x):
    return pd.Series([x.min(), x.max()], index=['min', 'max'])

frame.apply(defined_function)
````

💡 Por lo general no es necesario usar `apply`, pues varios de los métodos más comunes ya están implementados nativamente!

### `pandas.merge`

Corresponde a la manera de realizar lo que se conoce en `SQL` como joins. `pandas.merge()` Es la función utilizada por
los métodos `pandas.DataFrame.merge()` y `pandas.DataFrame.join()`

Pero antes, *qué es* un *merge*? un *join*?


Un `JOIN` se conoce como una operación de unión generalmente en el contexto de tablas de `SQL`. Hay varios tipos:

- INNER JOIN
- LEFT JOIN
- OUTER JOIN
- Otros (Existen distintos tipos de JOIN que dependen del tipo de motor SQL utilizado ej. Postgres, MySQL, etc)

Un JOIN se realiza utilizando una columna de la tabla como pivote

##### Una explicación visual (basado en https://blog.codinghorror.com/a-visual-explanation-of-sql-joins/)

Consideremos las siguientes tablas 

```
    Tabla A      Tabla B
id name       id  name
-- ----       --  ----
1  Pirate     1   Rutabaga
2  Monkey     2   Pirate
3  Ninja      3   Darth Vader
4  Spaghetti  4   Ninja
```

###### INNER JOIN

![venn_inner](figures/venn_inner_join.png)

Considera los elementos que resultan de la intersección de dos conjuntos de datos, **sólo** para aquellos valores que son iguales. Si hacemos un INNER JOIN entre ambas tablas, utilizando la columna `name` como pivote:

```
   Tabla A        Tabla B
1   Pirate     2    Pirate
3   Ninja      4    Ninja
```

Por qué resulta esta tabla?

###### FULL OUTER JOIN

![venn_inner](figures/venn_outer_join.png)

Resultado de la intersección de dos conjuntos; y para aquellos valores que no son iguales, la operación completa los vacíos con valores nulos. Para la columna `name`

```
id    name       id    name
-------------    -----------------
1     Pirate     2     Pirate
2     Monkey     null  null
3     Ninja      4     Ninja
4     Spaghetti  null  null
null  null       1     Rutabaga
null  null       3     Darth Vader
```

Por qué resulta esta tabla?

###### LEFT OUTER JOIN

![venn_inner](figures/venn_left_outer_join.png)

Produce la intersección de dos conjuntos de datos, y para aquellos valores que no son iguales, la operación completa los vacíos con valores nulos **sólo para la tabla izquierda**. 

Considerando la columna `name` como pivote:

```
id  name       id    name
--  ----       --    ----
1   Pirate     2     Pirate
2   Monkey     null  null
3   Ninja      4     Ninja
4   Spaghetti  null  null
```

¿Por qué resulta esta tabla?

`merge` combina dos dataframes utilizando una columna como pivote. Al igual que `concat` posee varios argumentos:

```python
pd.merge(left, right, how='inner', on=None, left_on=None, right_on=None,
         left_index=False, right_index=False, sort=True,
         suffixes=('_x', '_y'), copy=True, indicator=False,
         validate=None)
```


````python
left = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3'],
                      'A': ['A0', 'A1', 'A2', 'A3'],
                      'B': ['B0', 'B1', 'B2', 'B3']})
right = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3'],
                       'C': ['C0', 'C1', 'C2', 'C3'],
                       'D': ['D0', 'D1', 'D2', 'D3']})
result = pd.merge(left, right, on='key')
result
````
