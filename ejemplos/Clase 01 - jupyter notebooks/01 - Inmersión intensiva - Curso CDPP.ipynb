La listas **permiten almacenar cualquier tipo de valor como enteros, string y otro objetos**.Python tiene soporte para listas como un tipo predefinido del lenguaje. Para crear una lista basta con poner cosas entre [ ] (corchetes) y separarlas con , (comas).

````python
#Crear una lista vacía, (sin elementos), simplemente ponemos []
vacia=[]
print("Lista vacía:")
print(vacia)
````

````python
#Crear la lista de compras de la semana
print("-Lista de compras")
lista = ['Cochayuyo', 'Lechuga', 'Pan', 'Garbanzos','Tomate', 'Cebolla','Arroz']
print (lista)
````

### Diccionarios
Los diccionarios **permiten utilizar claves para declarar y acceder a un valor**. Se crean usando { } (llaves), donde cada elemento `(clave:valor)` se separa con una , (coma) dento de los corchetes.
Observa el siguiente ejemplo:

````python
notas_estudiantes = {'Felipe': 6.8, 'Manuel': 6.7, 'Raffaela': 7.0} 
print(notas_estudiantes)
````

**Los elementos son accedidos y modificados a través su índice**, tal como en las listas, exceptuando que **no hay un orden** que comience desde el cero, **sino que desde una clave presente en el diccionario**.

````python
#Acceder a un valor a través de su clave
notas_estudiantes['Manuel']
````

````python
#Agregar un nuevo par clave:valor al diccionario 'notas_estudiantes'
notas_estudiantes['Camila'] = 6.9 
print(notas_estudiantes)
````

**Nota:** Ten en cuenta que en un diccionario **el orden no es relevante**, fue diseñado pensando así. Esto permite que el acceso a sus elementos sea muy eficiente, independiente de su tamaño.

### If, elif, else
La sentencia condicional if se usa para tomar decisiones, este **evaluá básicamente una operación lógica**, es decir, **una expresión que de como resultado True o False**, y **ejecuta la pieza de código siguiente siempre y cuando el resultado sea verdadero.**

A continuación un de estructura condicional de if:

    if sentencia True:          
        pasa esto    
        
En el caso de que la sentencia sea `False`, se ejecuta `else`. Si hay más de dos opciones cuando la sentencia es `False`, se ejecuta `elif` y luego `else`.

  
  
  
Observa el siguiente ejemplo:

````python
#Chequear signo de número
x = -2

if x == 0:
    print("El número ingresado ("+str(x)+") es cero.")
elif x > 0:
    print("El número ingresado ("+str(x)+") es positivo.")
else:
    print("El número ingresado ("+str(x)+") es negativo.")
````

#### Conjuntos (Set)
La última colección básica de elementos es el conjunto o `set`, siendo una colección no ordenada de elementos únicos. Es mutable: una vez que se ha creado un conjunto, puede ser modificado. Se definen como las listas y tuplas, pero usan llaves como los diccionarios:

````python
primos = {2, 3, 5, 7}
impares = {1, 3, 5, 7, 9}
````

Si recordamos la notación y propiedades de los conjuntos en la mátemática, existen operaciones como la unión, intersección, diferencia, diferencia simétrica, entre otras. Python nos brinda la posibilidad de utilizar métodos u operadores para realizarlas:

````python
#Union: items appearing in either
primos | impares      # # con un operador
primos.union(impares) # equivalente con un método
````

````python
#Intersección: elementos que están en ambos grupos
primos & impares             # con un operador
primos.intersection(impares) # equivalente con método
````

````python
#Diferencia: elementos en primos pero no en impares
primos - impares           # con un operador
primos.difference(impares) # equivalente con método
````

````python
#Diferencia simétrica: elementos que aparecen en un solo conjunto
primos ^ impares                     # con un operador
primos.symmetric_difference(impares) # equivalente con método
````

#### For
El `for` es una sentencia de control que repite la condición establecida en un comienzo. A aquella sentencia que repita condiciones se le llama *bucle* o *ciclo*. Tiene la siguiente sintaxis:
```
for variable in lista:
    cuerpo del for
```

En este caso, `variable` es el objeto que va a ir cambiando.

````python
print("- Impresión de los elementos de la lista:")

#Imprimir los strings de mi_lista por separado

mi_lista=["img","python","numpy"]

for s in mi_lista:
        print(s, end=' ') # imprimir todo en la misma línea
````

````python
#Calcular la suma de los números e imprimirla

suma=0
pixel=[0.6,0.3,0.4]

for numero in pixel:
        suma+=numero # recordar que es lo mismo que suma = suma + numero

print("- La suma de los números es:")
print(round(suma,3))

print("- El promedio de ellos es:")
print(round(suma/len(pixel),3))
````

#### While 

La sentencia `while` permite repetir la ejecución de un grupo de instrucciones mientras se cumpla una condición (es decir, mientras la condición tenga el valor `True`). La sintaxis del ciclo while es la siguiente:

```
while condicion:
    cuerpo del ciclo
```
     
Python primero verifica la condición. Si es `False`, el ciclo finaliza y el control pasa a la siguiente instrucción después del cuerpo del ciclo while. Si la condición es `True`, se ejecuta el cuerpo del bucle y luego se vuelve a verificar la condición. Esto continúa mientras la condición sea verdadera Una vez que la condición se convierte en `False`, el ciclo termina y el control pasa a la siguiente declaración después del ciclo.

Por ejemplo, el siguiente fragmento de programa imprime los cuadrados de todos los enteros del 1 al 10. Aquí se puede reemplazar el ciclo "while" por el ciclo for ... in range (...):

````python
# Ejemplo 1 - while
#Mientras i sea menor que 10, elevar i al cuadrado
i=1
while i <= 10:
    print(i ** 2)
    i += 1
````

````python
# Ejemplo 2 - while
i = 1
while i <= 50:
    print('- Valor de i antes de cambiar:', i)
    i = 3*i + 1
    print('- Valor nuevo de i:', i)
print("Programa terminado")
````

````python
# Ejemplo 3 - while
numero = int(input("Escriba un número positivo: "))
while numero < 0:
    print("¡Ha escrito un número negativo! Inténtelo de nuevo") # se muestra el error
    numero = int(input("Escriba un número positivo: ")) # se vuelve a preguntar la condición inicial al final
print("¡Este sí es un número positivo!")
````

**Nota**: Si la condición de un ciclo se cumple siempre, no terminará nunca de ejecutarse y tendremos lo que se denomina un *ciclo infinito*. Los ciclos infinitos no intencionados deben evitarse, ya que implica perder el control del programa. Para interrumpirlo, hay que pulsar la combinación de teclas `Ctrl+C`.

````python
#Uso de While y else 
i = 1
while i <= 10:
    print(i)
    i += 1
else:
    print('Loop ended, i =', i)
````

## Desafio

Determinar si un número es primo o compuesto. Un número primo es aquel que **solo es divisible por uno o por si mismo.**
    
    Divisible: cuando el resto de división por un número dado es 0.
        Resto =  Dividendo  (divisor x cuociente)
        Dividendo - Resto = divisor x cuociente  

**Algoritmo**: Un número 𝑛 es primo si no podemos encontrar un número entero entre 2 y 𝑛−1 que lo haga divisible por dicho número.

 % Módulo: devuelve el resto cuando el primer operando se divide por el segundo

````python
#Solución 

#Un numero es primo cuando es divisible por 1 y por si mismo, es decir, cuando n%1=0 y n%n=0

n = int(input('Ingrese n: ')) #Entrega un numero
divisor = 2 #Primisa el divisor es 2
es_primo = True #Primisa es siempre primo

while (divisor != n): #Se corre el loop, mientras que el divisor (2) sea distinto del numero
    if (n % divisor == 0) : #Si el resto de la division es 0, el numero es par ...
        es_primo = False   #No es primo, es par
        divisor = n #el divisor es igual  al numero, se acaba el loop
        
    else: #caso contrario, si el resto es distinto de 0, es decir, si puede que sea un numero primo le sumamos uno al divisor
        divisor = divisor + 1 #le sumaremos uno al divisor mientras sea distinto que el numero 
if (es_primo == True) :
    print (n, 'es primo')
else:
    print (n, 'es compuesto')
````
